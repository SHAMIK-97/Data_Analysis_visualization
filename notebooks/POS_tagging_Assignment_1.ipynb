# Part-of-Speech (POS) Tagging

## What is POS tagging?

Part-of-speech tagging is the task of assigning a **grammatical category** (noun, verb, adjective, determiner, ...) to every word in a sentence. For example:

| Word | The | board | will | join | the | meeting | . |
|------|-----|-------|------|------|-----|---------|---|
| Tag  | DT  | NN    | MD   | VB   | DT  | NN      | . |

POS tagging is a basic building block of many NLP systems: it is used in parsing, named-entity recognition, information extraction, text-to-speech (e.g. to pronounce "record" as a noun or a verb) and machine translation.

## Why is it not trivial?

A simple dictionary lookup (word → tag) is not enough, because:

- **Ambiguity:** many words can take different tags depending on context. *"board"* is a noun in *"the board met"* but a verb in *"they board the plane"*; *"that"* can be a determiner, a pronoun or a conjunction.
- **Unknown words:** new words (names, numbers, technical terms) appear at test time that were never seen during training. The model must guess their tag from context and word form.
- **Fine-grained classes:** the Penn Treebank tagset distinguishes, for example, past tense (`VBD`) from past participle (`VBN`), which often look identical (*"joined"*) and can only be told apart from the surrounding words.

The tag of a word therefore depends on the **whole sentence**, not only on the word itself.

## How we approach it

POS tagging is a **sequence labelling** problem, a *many-to-many* task with aligned input and output: the model reads a sequence of *N* words and outputs a sequence of *N* tags, one per word. Recurrent neural networks fit this naturally, because they process the sentence word by word while carrying information about the context.

In this notebook we:

1. Convert words into integer ids and then into dense vectors (word embeddings).
2. Train several recurrent architectures that predict one tag per word, such as a **baseline LSTM**, a **GRU**, a **bidirectional LSTM** (which also sees the words to the right of each position), and deeper variants with stacked recurrent or dense layers.
3. Compare the models on held-out data.

## How performance is measured

The tag distribution is highly imbalanced (nouns, prepositions and determiners are very frequent, while some tags occur only a few times). Plain accuracy would reward a model that handles the frequent tags and ignores the rare ones. We therefore use the **macro-averaged F1 score**, which gives every tag the same weight, and inspect **confusion matrices** to see which tags the models confuse. Padding positions, and typically punctuation tags, are excluded from the evaluation.

# The Dataset: Penn Treebank sample (`dependency_treebank`)

## Source

We use the sample of the **Penn Treebank** distributed with NLTK under the name `dependency_treebank`. It contains articles from the **Wall Street Journal** (1989), so the language is formal news text with many company names, numbers and financial terms. In total it has roughly 3,900 sentences and about 94,000 tokens.

## How the data is arranged

After downloading and extracting the zip archive, the data is stored as **199 plain-text files**, one per news article:

    Datasets/
    └── dependency_treebank/
        ├── wsj_0001.dp
        ├── wsj_0002.dp
        ├── ...
        └── wsj_0199.dp

Inside each `.dp` file:

- **Each line is one token** (a word or a punctuation mark).
- **Fields on a line are separated by a tab character.**
- **Sentences are separated by a blank line.**

Example (the first sentence of `wsj_0001.dp`):

    Pierre      NNP   2
    Vinken      NNP   8
    ,           ,     2
    61          CD    5
    years       NNS   6
    old         JJ    2
    ,           ,     2
    will        MD    0
    join        VB    8
    ...

## Columns

Each line has **3 columns**, which we name `word`, `tag` and `value` when loading:

| Column | Content | Example | Used for POS tagging? |
|--------|---------|---------|-----------------------|
| `word` | The token as it appears in the text: a word, a number, or a punctuation mark. Abbreviations keep their period (e.g. `Nov.`). | `Vinken` | **Yes — input (X)** |
| `tag` | Its Penn Treebank POS tag. There are 45 distinct tags in the data, including one tag per punctuation type. | `NNP` | **Yes — target (y)** |
| `value` | Position (1-based, within the sentence) of the token's syntactic **head** in the dependency tree; `0` marks the root of the sentence (usually the main verb). | `2` | **No** — this belongs to the dependency-parsing task and is discarded |

## Tagset overview

The 45 tags fall into a few groups:

- **Nouns:** `NN` (singular), `NNS` (plural), `NNP` (proper, singular), `NNPS` (proper, plural)
- **Verbs:** `VB` (base form), `VBD` (past tense), `VBG` (gerund / present participle), `VBN` (past participle), `VBP` (present, non-3rd person), `VBZ` (present, 3rd person singular), `MD` (modal)
- **Adjectives and adverbs:** `JJ`, `JJR`, `JJS`, `RB`, `RBR`, `RBS`
- **Function words:** `DT` (determiner), `IN` (preposition / subordinating conjunction), `CC` (coordinating conjunction), `TO`, `PRP` / `PRP$` (pronouns), `WDT`, `WP`, `WP$`, `WRB` (wh-words), `EX`, `PDT`, `POS`, `RP`
- **Other:** `CD` (number), `FW` (foreign word), `UH` (interjection), `SYM`, `LS`
- **Punctuation and symbols:** `,` `.` `:` `` `` `` `''` `-LRB-` `-RRB-` `$` `#`

## Train / validation / test split

The split is done **by file**, so that all sentences of one article stay in the same split:

| Split | Files | Purpose |
|-------|-------|---------|
| Training | `wsj_0001` – `wsj_0100` | Fit the model weights |
| Validation | `wsj_0101` – `wsj_0150` | Choose hyperparameters, early stopping, select the best model |
| Test | `wsj_0151` – `wsj_0199` | Final, one-time comparison of the models |

## What we have to do to accomplish the task

1. **Download and extract** the archive into the `Datasets` folder.
2. **Read each file** and keep only the `word` and `tag` columns, making sure every token is read as literal text (so words such as "NA" are not turned into missing values).
3. **Rebuild the sentences** so each training example is a pair (list of words, list of tags) of equal length.
4. **Split** the sentences into training, validation and test sets by file number.
5. **Explore the data:** split sizes, tag distribution (class imbalance), sentence lengths.
6. **Build vocabularies:** map every word and every tag to an integer id, with id 0 reserved for padding and a special id for out-of-vocabulary words.
7. **Encode and pad:** convert sentences and tag sequences into integer sequences and pad them to a common length, so they can be processed in batches.
8. **Represent words as vectors** with an embedding layer (randomly initialised or pre-trained, e.g. GloVe).
9. **Train several recurrent models** (LSTM, GRU, bidirectional, stacked variants) that output one tag per word, using class weights and early stopping on the validation set.
10. **Evaluate on the test set** with macro-averaged F1, excluding padding (and usually punctuation), and analyse the errors with confusion matrices.
11. **Compare the models** and discuss which architecture works best and why.

##Preparing the Train validation test split from the downloaded corpus

# spliting documents 1-100 are the train set, 101-150 validation set, 151-199 test set.

Splitting is done in sentences



```python
# ============================================================================
# READ THE CORPUS AND SPLIT IT INTO TRAIN / VALIDATION / TEST
# Reads each .dp file from the zip archive, cuts the token stream into
# sentences, and assigns each file's sentences to one split:
#   files   1-100 -> training
#   files 101-150 -> validation
#   files 151-199 -> test
# Splitting by file keeps sentences from the same article in one split.
# ============================================================================

# List every entry in the zip archive (the "dependency_treebank/" folder
# entry plus the 199 .dp files) and sort it. Zero-padded names make
# alphabetical order equal to numeric order, and the folder entry sorts
# first (index 0).
files = zipAll.namelist()
files.sort()
print('No. of files: {}'.format(len(files)))


def split_sentences(sentences, tags):
    """Cut a flat token list (and its parallel tag list) into sentences,
    ending a sentence each time the token '.' appears."""
    # Completed sentences and their tag sequences.
    final_sentences, final_tags = [], []
    # Buffers for the sentence currently being built.
    ret_sentences, ret_tags = [], []
    for i in range(0, len(sentences)):
        # Add the current token and its tag to the current sentence.
        ret_sentences.append(sentences[i])
        ret_tags.append(tags[i])
        # A period closes the sentence (and is kept as its last token).
        if sentences[i] == '.':
            final_sentences.append(ret_sentences)
            final_tags.append(ret_tags)
            # Start fresh buffers for the next sentence.
            ret_sentences, ret_tags = [], []
    # Keep any leftover tokens as a final sentence without a period.
    if len(ret_sentences) > 0:
        final_sentences.append(ret_sentences)
        final_tags.append(ret_tags)
    return (final_sentences, final_tags)


# Quick test on a hand-made example: expect two sentences, split after '.'.
x, y = split_sentences(['The','best','run','of','all','.','It','is','running','for','you'],
                       ['DT','JJS','VB','IN','DT','.','PRP','VB','VBG','IN','PRP'])
print('x:', x)
print('y:', y)

# One list of sentences and one parallel list of tag sequences per split.
train_sentences, train_tags = [], []
validation_sentences, validation_tags = [], []
test_sentences, test_tags = [], []

# Loop over the archive entries, starting at 1 to skip the folder entry.
# f = 1 is wsj_0001.dp, ..., f = 199 is wsj_0199.dp.
for f in range(1, len(files)):
    # Read one .dp file directly from the archive:
    # - header=None: no header row in the file.
    # - delimiter='\t': fields are tab-separated.
    # - names=[...]: word, tag, and the head index ('value', unused).
    # - keep_default_na=False: keep tokens like "NA" or "null" as text
    #   instead of converting them to NaN (this caused the tokenizer error).
    # - quoting=3 (csv.QUOTE_NONE): treat quote characters as plain text.
    # - dtype={'word': str, 'tag': str}: always read words and tags as
    #   strings, so numeric tokens like "1987" stay text.
    my_df = pd.read_csv(zipAll.open(files[f]), header=None,
                        delimiter='\t', names=['word', 'tag', 'value'],
                        keep_default_na=False, quoting=3,
                        dtype={'word': str, 'tag': str})

    # Split this file's token stream into sentences with aligned tags.
    x, y = split_sentences(my_df['word'].to_numpy(), my_df['tag'].to_numpy())

    # Assign all sentences of this file to one split according to its number.
    if f <= 100:
        train_sentences.extend(x)
        train_tags.extend(y)
    elif f <= 150:
        validation_sentences.extend(x)
        validation_tags.extend(y)
    else:
        test_sentences.extend(x)
        test_tags.extend(y)

# Safety check: every word and every tag must be a string. A NaN here would
# mean a real token was turned into a missing value, which breaks the
# tokenizer later.
assert all(isinstance(tok, str)
           for sents in (train_sentences, validation_sentences, test_sentences)
           for sent in sents for tok in sent), "NaN token found in the sentences"
assert all(isinstance(tag, str)
           for tags in (train_tags, validation_tags, test_tags)
           for seq in tags for tag in seq), "NaN tag found in the tags"

# Sentence counts per split and in total.
number_training_sentences = len(train_sentences)
number_validation_sentences = len(validation_sentences)
number_test_sentences = len(test_sentences)
number_sentences = number_training_sentences + number_validation_sentences + number_test_sentences
# Number of tag sequences; must equal the number of sentences.
number_tags = len(train_tags) + len(validation_tags) + len(test_tags)

print('No. sentences: {}'.format(number_sentences))
print('No. tag sentences: {}'.format(number_tags))
print('No. Training sentences: {}'.format(number_training_sentences))
print('No. Validation sentences: {}'.format(number_validation_sentences))
print('No. Test sentences: {}'.format(number_test_sentences))
print("Successful Split")
```

## From text encoding to vectorization

### 1. Text encoding is not vectorization

The tokenization step **encoded** the text: every word was replaced by an integer id from the vocabulary.

    "the board will join"  →  [3, 347, 52, 1204]

These ids are only **labels**. They tell us *which* word appears, but nothing *about* the word:

- The numeric values have no meaning: id 500 is not "larger", "closer" or "more similar" to id 501 than to id 20.
- Nothing in the ids shows that *"shares"* and *"stocks"* are used alike, or that *"joined"* and *"entered"* are both past-tense verbs.
- A neural network that received raw ids as numbers would treat these arbitrary values as meaningful quantities, which would mislead it.

**Vectorization** is the next step: replacing each id by a **vector of numbers that describes the word**, so that words used in similar ways get similar vectors. The encoding is the index; the vectorization is the content we look up with that index.

    "board"  →  id 347           →  [0.12, -0.53, 0.77, ..., 0.08]
                (encoding:          (vectorization:
                 which word)         what the word is like)

### 2. Ways to vectorize words

**a) One-hot vectors**

Each word becomes a vector of length |V| (9,903) with a single 1 at its id and 0 everywhere else.

- Simple, and removes the false ordering of the ids.
- But every word is equally different from every other word (no similarity), and the vectors are huge and almost entirely zeros.

**b) Rows of the co-occurrence matrix (sparse, count-based)**

Based on the **distributional hypothesis**: *a word is characterised by the words around it*. Row *i* of the co-occurrence matrix counts how often each vocabulary word appears within the context window of word *i*, so it can be used directly as the vector of word *i*.

- Words used in similar contexts get similar rows: two nouns both frequently follow *"the"* and *"a"*.
- But the vectors are still |V|-dimensional and extremely sparse (0.09 % non-zero in our matrix), raw counts are dominated by frequent words (our largest count is the pair *the*–*of*), and rare words have almost empty, noisy rows, since our corpus is small.
- Common improvements: reweight counts with **PPMI** (positive pointwise mutual information), which highlights pairs that co-occur more than their frequencies alone would predict, and compress the matrix with **SVD** to obtain short dense vectors.

**c) Pre-trained dense embeddings (GloVe)**

GloVe applies the same co-occurrence idea **at a much larger scale** and **compresses** the result:

- It is trained on the global co-occurrence statistics of billions of tokens (Wikipedia and Gigaword news).
- It learns a dense vector of 50–300 dimensions for each of its 400,000 words, such that the vectors reflect how often words co-occur.
- Similar words get nearby vectors, and the vectors capture syntactic regularities (nouns near nouns, past tenses near past tenses) as well as semantic ones.

In other words, GloVe gives what our co-occurrence rows aim for (vectors that reflect word usage), but in compact form and estimated from far more text than we have, so even words that are rare in our data get reliable vectors.

**d) Embeddings learned during training**

An **Embedding layer** can also start from random vectors and learn them while training the tagger. This works, but with only about 2,000 training sentences, most words are seen too rarely to learn good vectors. Initialising the layer with GloVe and optionally fine-tuning it combines both approaches.

### 3. How vectorization is done in this notebook

1. **Encode:** each word → its id (done by the tokenizer).
2. **Build an embedding matrix** of shape |V| × *d* (e.g. 9,903 × 100), where row *i* is the vector for word id *i*:
   - words found in GloVe → their GloVe vector;
   - words not in GloVe (OOV with respect to GloVe) → a random vector (or zeros);
   - `<PAD>` (id 0) → a zero vector.
3. **Load this matrix into the Embedding layer.** During training and prediction, the layer simply **looks up row *i* for every id *i*** in the input sequence.
4. The recurrent layers (LSTM, GRU, ...) read these vectors **in order** and predict one tag per word.

Mathematically, the lookup is the same as multiplying the one-hot vector of a word by the embedding matrix, which is why embeddings can be seen as a compressed, learned replacement for one-hot vectors.

### 4. Why this is not TF-IDF

TF-IDF also starts from the vocabulary and counts words, but it counts different things:

| | TF-IDF | Co-occurrence matrix / GloVe |
|---|---|---|
| What is counted | How often a word appears in each **document** | How often a word appears **next to each other word** |
| Matrix | word × document | word × word |
| A vector describes | A document's content (its topic) | A word's usage (its contexts) |
| Word order / position | Ignored | Local context captured |
| Output | One vector per document | One vector per word |
| Typical use | Search, document classification | Word embeddings, sequence labelling |

POS tagging needs **one vector per word** that reflects **how the word is used**, so it relies on word-context representations (GloVe), not on document-level representations (TF-IDF). The two share one issue: raw counts are dominated by frequent words. TF-IDF corrects this with the IDF weight, and co-occurrence methods with PPMI or with the weighting built into GloVe's training.

### Summary

| Step | Output | Carries meaning? |
|------|--------|------------------|
| Encoding (tokenizer) | Integer id per word | No, only identity |
| One-hot | Sparse |V|-dim vector | No similarity between words |
| Co-occurrence rows | Sparse |V|-dim count vector | Yes, but noisy and huge |
| GloVe embedding | Dense 50–300-dim vector | Yes, compact and reliable |

## Tokenization of Data

The next thing we need to figure out is how are we going to feed these inputs to an RNN. If we have to give the words as input to any neural networks then we essentially have to convert them into numbers. We need to create a word embedding or one-hot vectors i.e. a vector of numbers for each word. Therefor, we'll first encode the input and output which will give a blind unique id to each word in the entire corpus for training data and then incrementaly build upon it so that our model can be well utilized in the real world scenarios. On the other hand, we have the Y matrix(tags/output data). We have forty-five POS tags here, treating each of them as a class and each pos tag is converted into one-hot encoding.

----

### How our tokenization relates to Bag-of-Words and TF-IDF

Bag-of-Words (BoW), TF-IDF and our approach all start with the same step: **building a vocabulary** that maps every distinct token to an integer id. What differs is what is done with that vocabulary afterwards, and that difference decides which method suits POS tagging.

**1. The shared first step: the vocabulary**

In all three cases, the text is split into tokens and each distinct token gets an index. Our `word_to_idx` dictionary (9,903 entries) is exactly the kind of vocabulary that scikit-learn's `CountVectorizer` or `TfidfVectorizer` build internally. So **tokenization and vocabulary building are the same idea**; the methods diverge in how they turn text into vectors.

**2. Bag-of-Words: counts per document, order discarded**

BoW represents a whole **document (or sentence) as a single vector** of length *V* (the vocabulary size), where position *i* holds how many times word *i* occurs.

- *"the board will join the meeting"* → a 9,903-dimensional vector with 2 at the position of *"the"*, 1 at *"board"*, *"will"*, *"join"*, *"meeting"*, and 0 everywhere else.
- The **word order is lost**: *"dog bites man"* and *"man bites dog"* get the same vector.

**3. TF-IDF: weighted Bag-of-Words**

TF-IDF keeps the BoW structure (one vector per document, no order) but **reweights** the counts:

- **TF** (term frequency): how often the word appears in this document.
- **IDF** (inverse document frequency): lower weight for words that appear in many documents (*"the"*, *"of"*, *"is"*), higher weight for rare, distinctive words.

The idea is that frequent function words carry little information about **what a document is about**, which makes TF-IDF effective for document classification, search and topic analysis.

**4. What POS tagging needs instead**

POS tagging assigns a label to **every word**, and the correct label depends on the **surrounding words in order**. BoW and TF-IDF fail on both counts:

- They produce **one vector per sentence**, but we need **one prediction per token**.
- They **discard word order**, yet order is exactly what disambiguates tags: *"the **board** met"* (NN) vs *"they **board** the plane"* (VBP).
- TF-IDF **down-weights function words**, while for tagging these words are among the most informative: a determiner (*"the"*) strongly signals that a noun or adjective follows, and a modal (*"will"*) that a base-form verb (VB) follows.

Therefore we do not turn a sentence into a single vector. We turn it into a **sequence of integer ids**, one per token, in the original order:

    "Pierre Vinken , 61 years old"  →  [id(pierre), id(vinken), id(,), id(61), id(years), id(old)]

**5. From ids to vectors: one-hot and embeddings**

An integer id is only a label; its numeric value means nothing (id 500 is not "bigger" than id 20). Each id must become a vector:

- **One-hot encoding:** a vector of length *V* with a single 1 at the word's id. This is literally the BoW vector of a one-word document. It is sparse (9,903 dimensions, all but one equal to 0) and treats all words as equally different from each other: *"good"* is as far from *"great"* as from *"table"*.
- **Word embeddings** (what our models use): the Embedding layer maps each id to a **dense, low-dimensional vector** (typically 50–300 dimensions) that is learned during training, or initialised from pre-trained embeddings such as GloVe. Words used in similar contexts get similar vectors, so the model can generalise, e.g. from *"joined"* to *"entered"*. Mathematically, an embedding lookup equals multiplying the one-hot vector by a learned weight matrix, so embeddings are a compressed, learned version of one-hot vectors.

The recurrent layers (LSTM, GRU) then read these embedding vectors **in order**, which provides the context that BoW and TF-IDF throw away.

**Summary**

| | Bag-of-Words | TF-IDF | Our approach (ids + embeddings) |
|---|---|---|---|
| Vocabulary / tokenization | Yes | Yes | Yes |
| Unit represented | Whole document | Whole document | Each token |
| Word order | Lost | Lost | Kept |
| Vector type | Sparse counts, size *V* | Sparse weights, size *V* | Dense, learned, small size |
| Word similarity captured | No | No | Yes |
| Function words | Counted like any word | Down-weighted | Fully used as context |
| Typical use | Document classification | Search, document classification | Sequence labelling (POS, NER), language models |

**Could BoW-style features be used for tagging at all?** Yes, in a limited way. Classical, pre-neural taggers classify each word using hand-built features of a **fixed window** around it (e.g. the previous and next two words, suffixes, capitalization), which is a local, per-token bag of features. This works reasonably well, but the window is fixed and the features are designed by hand. Recurrent networks over embeddings learn which context matters, and bidirectional models can use the whole sentence on both sides of each word.

---
### What is OOV (Out-Of-Vocabulary)?

An **out-of-vocabulary (OOV) word** is a word that is **not in the vocabulary** the model was built with. The vocabulary is fixed when we fit the tokenizer (here, on the training and validation sentences), so any word that appears later, for example in the test set or in real-world text, but was never seen during fitting has no id of its own.

**Why OOV words are unavoidable**

Language is open-ended. However large the training corpus, new text always contains words it did not include:

- **Proper names:** people, companies, places (*"Vinken"*, a new company name). Very frequent in news text.
- **Numbers and dates:** *"1,453"*, *"29"*, *"$4.7"* — almost every specific number is rare.
- **Rare or technical words:** *"nonexecutive"*, domain-specific terms.
- **New words, spelling variants and typos.**

Word frequencies follow a very skewed distribution (Zipf's law): a few words are extremely common, while a large fraction of the distinct words appear only once. So even a test set from the same newspaper contains a noticeable share of unseen words.

**How OOV words are handled in this notebook**

The tokenizer reserves a special token, `--oov--`, with **id 1**. When a sentence is encoded:

- known words are replaced by their own id;
- every unknown word is replaced by id 1.

    Vocabulary:   {"<PAD>": 0, "--oov--": 1, "the": 2, "board": 3, ...}
    Test input:   "the Zorblax board"
    Encoded:      [2, 1, 3]        ← "zorblax" is unknown → id 1

Without an OOV token, unknown words would either be **dropped**, which would shift every following word out of alignment with its tag, or cause an error. Mapping them to a shared id keeps the sequence length intact, so each word still gets exactly one predicted tag.

**How the model can still tag an unknown word**

All OOV words share one embedding vector, so the model cannot tell them apart by their identity. It must rely on **context**: in *"the Zorblax board"*, the determiner before and the noun after make an adjective or proper noun likely. Recurrent models, especially bidirectional ones that see both sides of the word, are well suited for this. Accuracy on OOV words is nevertheless typically much lower than on known words, which makes OOV handling one of the main sources of tagging errors.

**An important detail for our setup**

Because the vocabulary was built from the training and validation sentences, **no training word is ever mapped to `--oov--`**. The embedding row for id 1 therefore never receives a gradient update and stays at its initial random value, while at test time every unknown word is mapped to it. Common remedies:

- **Replace rare training words with `--oov--`** (e.g. words occurring only once), so the model learns what an "unknown word" looks like in context.
- **Randomly replace a small fraction of training words with `--oov--`** during training (word dropout), with the same goal.
- **Use pre-trained embeddings** (e.g. GloVe): a word unseen in our corpus may still have a meaningful pre-trained vector, reducing the number of true OOV words.
- **Add word-form information**: capitalization, suffixes (*-ing*, *-ed*, *-ly*), or character-level / subword representations let the model guess the tag of an unknown word from its shape (*"blorfing"* is probably VBG).

```python
# ============================================================================
# MINIMAL TOKENIZER (replacement for keras.preprocessing.text.Tokenizer)
# keras.preprocessing.text was removed in Keras 3, which is now the default
# in TensorFlow and on Colab. This small class reproduces the part of the
# old Keras Tokenizer that this notebook uses, with the same attribute
# names, so the rest of the cell works unchanged:
#   - word_counts: token -> number of occurrences, in first-seen order
#   - word_index:  token -> id; OOV token at 1 (if given), then tokens by
#                  decreasing frequency; id 0 is never assigned (padding)
#   - index_word:  id -> token (inverse of word_index)
#   - texts_to_sequences / sequences_to_texts: encode and decode
# ============================================================================

# Counter counts hashable items; since Python 3.7 it remembers the order in
# which items were first inserted, which is used to break frequency ties.
from collections import Counter


class Tokenizer:
    """Maps tokens to integer ids. Input texts are lists of tokens
    (already split into words), as in this notebook; plain strings are
    also accepted and split on whitespace."""

    def __init__(self, oov_token=None, lower=True):
        # Token used for words not seen during fitting. None means "no OOV
        # entry", which is the right choice for the closed set of POS tags.
        self.oov_token = oov_token
        # Whether to lowercase every token (Keras' default is True).
        self.lower = lower
        # Occurrence counts, filled by fit_on_texts.
        self.word_counts = Counter()
        # Forward and inverse vocabularies, filled by fit_on_texts.
        self.word_index = {}
        self.index_word = {}

    def _prepare(self, text):
        """Turn one text into a list of tokens, lowercased if required."""
        # A plain string is split on whitespace; a list is used as it is,
        # so punctuation tokens such as ',' and '.' are kept intact.
        tokens = text.split() if isinstance(text, str) else list(text)
        # Lowercase each token when lower=True ("The" -> "the").
        return [t.lower() for t in tokens] if self.lower else tokens

    def fit_on_texts(self, texts):
        """Count all tokens in `texts` and assign ids by frequency."""
        # Update the counts with every token of every text. Calling
        # fit_on_texts again adds to the existing counts, as in Keras.
        for text in texts:
            self.word_counts.update(self._prepare(text))
        # Sort tokens by decreasing count. Python's sort is stable, so
        # tokens with equal counts keep their first-seen order, which is
        # the same tie-breaking rule as the Keras Tokenizer.
        sorted_tokens = sorted(self.word_counts, key=self.word_counts.get,
                               reverse=True)
        # Put the OOV token first if one was requested, so it gets id 1.
        if self.oov_token is not None:
            sorted_tokens = [self.oov_token] + sorted_tokens
        # Assign ids starting at 1: id 0 stays free for padding.
        self.word_index = {tok: i for i, tok in enumerate(sorted_tokens, start=1)}
        # Inverse mapping id -> token, for decoding.
        self.index_word = {i: tok for tok, i in self.word_index.items()}

    def texts_to_sequences(self, texts):
        """Convert each text into a list of ids."""
        # Id of the OOV token, or None if the tokenizer has no OOV entry.
        oov_id = self.word_index.get(self.oov_token)
        sequences = []
        for text in texts:
            seq = []
            for tok in self._prepare(text):
                if tok in self.word_index:
                    # Known token: use its id.
                    seq.append(self.word_index[tok])
                elif oov_id is not None:
                    # Unknown token with an OOV entry: map it to the OOV id.
                    seq.append(oov_id)
                else:
                    # Unknown token and no OOV entry. The old Keras
                    # Tokenizer silently DROPPED such tokens, which would
                    # shift every following tag relative to its word.
                    # Raising an error here makes the problem visible.
                    raise KeyError(f"Unknown token {tok!r} and no oov_token set")
            sequences.append(seq)
        return sequences

    def sequences_to_texts(self, sequences):
        """Convert lists of ids back into lists of tokens, skipping the
        padding id 0 (which has no token in index_word)."""
        return [[self.index_word[i] for i in seq if i in self.index_word]
                for seq in sequences]
```

```python
# ============================================================================
# TOKENIZATION: BUILD THE WORD AND TAG VOCABULARIES
# Neural networks work on numbers, not strings. This cell assigns an
# integer id to every distinct word and every distinct POS tag, and builds
# lookup dictionaries in both directions (word -> id, id -> word).
# In later cells these ids turn each sentence into an integer sequence
# (input X) and each tag sequence into class indices (output Y), which are
# then padded to a common length.
#
# Id layout after this cell:
#   words: 0 = <PAD> (padding), 1 = --oov-- (unknown word), 2.. = real words
#   tags:  0 = <PAD> (padding),                          1.. = real tags
# Id 0 is used ONLY for padding. This matters because pad_sequences fills
# short sentences with 0, and an Embedding layer with mask_zero=True
# ignores positions with id 0. No real word or tag may share that id.
#
# The word vocabulary is fitted on training + validation sentences; the
# test sentences are excluded so that test words remain unseen until the
# final evaluation. The tag vocabulary is fitted on the training tags.
# ============================================================================

# tqdm wraps an iterable and shows a progress bar while looping over it.
# Used in evaluate_vocabulary to show progress over the vocabulary entries.
from tqdm import tqdm

# TensorFlow, the backend that runs the Keras models. Not used directly in
# this cell; kept here for the model-building cells that follow.
import tensorflow as tf

# Special token used to represent padding positions in the word
# dictionaries. It is written in uppercase so it can never collide with a
# real token: the word tokenizer lowercases every word it sees.
PAD_TOKEN = '<PAD>'

# Special token for out-of-vocabulary words (words not seen when fitting).
OOV_TOKEN = '--oov--'


def words_tokenizer(sentences, tags):
    """Fit one Tokenizer on the sentences and one on the tags, and build
    word <-> id dictionaries with id 0 reserved for padding.
    `sentences` and `tags` are lists of token lists (already split into
    words), not raw strings."""

    # ---------------------------------------------------------------- words
    # Create the word tokenizer:
    # - oov_token: adds an "unknown word" entry at id 1. Any word not seen
    #   during fitting (e.g. a new word in the test set) is mapped to this
    #   id instead of being silently dropped from the sequence.
    # - lower=True (the default, written explicitly for clarity): every word
    #   is lowercased, so "The" and "the" share one id. This shrinks the
    #   vocabulary and matches lowercased pre-trained embeddings such as
    #   GloVe, at the cost of losing capitalization, a useful cue for proper
    #   nouns (NNP). Set lower=False to keep case.
    # Punctuation filters are NOT applied when the input is a list of
    # tokens, so tokens like ',' and '.' are kept; they have their own tags.
    word_tokenizer = Tokenizer(oov_token=OOV_TOKEN, lower=True)

    # Scan all sentences and count every (lowercased) word. After fitting:
    # - word_counts: word -> number of occurrences (without the OOV token).
    # - word_index:  word -> id; OOV at 1, then words by decreasing
    #   frequency from id 2 up to V+1, where V is the number of distinct
    #   words. Id 0 is left unused.
    word_tokenizer.fit_on_texts(sentences)

    # Plain dict of word counts (word -> count), for later analysis.
    word_count = dict(word_tokenizer.word_counts)

    # Alphabetically sorted list of all distinct words, i.e. the vocabulary
    # without any special token. Its length is V.
    word_listing = sorted(word_count.keys())

    # Build the word -> id dictionary as a NEW dict:
    # - {PAD_TOKEN: 0} puts the padding token at id 0;
    # - **word_tokenizer.word_index copies the tokenizer's entries
    #   (OOV at 1, real words from 2 upward) without changing their ids.
    # Because this is a copy, the tokenizer itself is left untouched, so
    # word_tokenizer.texts_to_sequences(...) produces exactly the ids found
    # in this dictionary, and id 0 never appears for a real word.
    word_to_idx = {PAD_TOKEN: 0, **word_tokenizer.word_index}

    # Inverse mapping id -> word, built by swapping each (word, id) pair.
    # Used to turn encoded or predicted sequences back into words.
    idx_to_word = {idx: word for word, idx in word_to_idx.items()}

    # ----------------------------------------------------------------- tags
    # Create the tag tokenizer:
    # - no oov_token: the tag set is closed (the ~45 Penn Treebank tags all
    #   occur in the training data), so an "unknown tag" class would only
    #   add an output unit that is never used.
    # - lower=False: keep the original tag names ('NNP', not 'nnp') so that
    #   classification reports and confusion matrices show the standard
    #   Penn Treebank labels.
    # Without an OOV entry, the tags get ids 1..T by decreasing frequency,
    # and id 0 stays free for padding.
    tag_tokenizer = Tokenizer(lower=False)

    # Fit on the tag sequences: each distinct tag receives its id.
    tag_tokenizer.fit_on_texts(tags)

    # Tag -> id dictionary with padding at 0, built as a copy (same reason
    # as for the words: the tokenizer must stay consistent with it).
    tag_to_idx = {PAD_TOKEN: 0, **tag_tokenizer.word_index}

    # Inverse mapping id -> tag, needed to decode predictions and to label
    # the axes of the confusion matrix.
    idx_to_tag = {idx: tag for tag, idx in tag_to_idx.items()}

    # Return the fitted tokenizers, the sorted word list, and the four
    # lookup dictionaries.
    return (word_tokenizer, tag_tokenizer, word_listing,
            word_to_idx, idx_to_word, tag_to_idx, idx_to_tag)


def evaluate_vocabulary(idx_to_word, word_to_idx, word_listing, n_special):
    """Check that the two dictionaries are exact inverses, that the ids
    form the contiguous range 0..N-1, and that id 0 is the padding token.
    n_special is the number of special tokens added on top of the real
    entries (2 for words: <PAD> and --oov--; 1 for tags: <PAD>)."""

    # Check size
    print("[Vocabulary Evaluation] Size checking...")

    # Both directions must have the same number of entries. If two entries
    # shared an id, idx_to_word would be smaller (dict keys are unique).
    assert len(idx_to_word) == len(word_to_idx)

    # The dictionaries hold every real entry plus the special tokens.
    assert len(idx_to_word) == len(word_listing) + n_special

    # Padding must occupy id 0, and no other entry may use it.
    assert idx_to_word[0] == PAD_TOKEN

    # Check content
    print("[Vocabulary Evaluation] Content checking...")

    # Loop over every id from 0 to N-1, with a progress bar.
    for i in tqdm(range(0, len(idx_to_word))):
        # Every id must exist (a gap raises KeyError) and map to an entry
        # that is also present in the forward dictionary.
        assert idx_to_word[i] in word_to_idx
        # Round trip: id -> entry -> id must return the same id.
        assert word_to_idx[idx_to_word[i]] == i


# Build the vocabularies:
# - words: fitted on training + validation sentences (list(...) makes
#   copies, and + concatenates them into one list of sentences);
# - tags: fitted on the training tags only.
(word_tokenizer, tag_tokenizer, word_listing,
 word_to_idx, idx_to_word, tag_to_idx, idx_to_tag) = words_tokenizer(
    list(train_sentences) + list(validation_sentences), list(train_tags))

# Run the consistency checks on the word vocabulary (2 special tokens:
# padding and OOV). A failed assert stops the cell with an AssertionError.
print("Vocabulary evaluation...")
evaluate_vocabulary(idx_to_word, word_to_idx, word_listing, n_special=2)

# Same checks on the tag vocabulary (1 special token: padding). The list
# of real tags plays the role of word_listing.
print("Tag vocabulary evaluation...")
evaluate_vocabulary(idx_to_tag, tag_to_idx,
                    sorted(tag_tokenizer.word_index.keys()), n_special=1)
print("Evaluation completed!")

# Sizes needed when building the models:
# - vocabulary size for the Embedding layer's input_dim (padding + OOV +
#   words), so every possible id has a row in the embedding matrix;
# - number of output classes for the final Dense/softmax layer (padding +
#   real tags). The padding class is later masked out of the loss and the
#   metrics.
vocab_size = len(word_to_idx)
num_tags = len(tag_to_idx)
print(f"Vocabulary size (incl. <PAD> and OOV): {vocab_size}")
print(f"Number of tag classes (incl. <PAD>): {num_tags}")
```

## Building the Co-occurence Matrix

As known, embedding methods are based on the principle that similar words will be used in similar contexts. Thus, context information is crucial to determine the meaning of a word.

One basic approach, which falls under the category of sparse representations, is the **co-occurrence matrix**: for each word in the vocabulary we count the number of times each other word appears within the same context window. A simple example is given by image below.

![](https://drive.google.com/uc?export=view&id=1UknGoYvIBBA7ytkSlqm1NhF_lHt0iOwT)

In particular, the context window defines our notion of word context. Consider the following example:

<h3><center> The cat is on the table </center></h3>

We have to consider each word in the sentence and for each we have count words within the context window. Suppose a window of size 2, then we have for the word 'cat':

Current word: cat

Context words: [the, is, on]

Notice how we consider $W$ words back and ahead of current word, where $W$ is the window size.




Let's define the simplest version of a **co-occurrence matrix** based on word counting.



```python
# ============================================================================
# CO-OCCURRENCE MATRIX
# Builds a |V| x |V| word-word matrix where cell (i, j) counts how often
# word j appears within the context window of word i, over the training +
# validation sentences. Each row is a sparse, count-based vector for one
# word: words used in similar contexts get similar rows, which is the idea
# behind embeddings. The matrix is then checked for consistency.
#
# Context definition (as in the markdown above): for each word at position
# p, the context is the up-to-W words before it and the up-to-W words
# after it, where W is the window size. Each (word, context word) pair is
# counted once per occurrence.
# ============================================================================

# scipy.sparse provides matrix formats that store only the non-zero
# entries. With |V| = 9,903 the full matrix has ~98 million cells, but most
# word pairs never co-occur, so a sparse matrix needs only a small fraction
# of the memory of a dense NumPy array.
import scipy.sparse

# gc: interface to Python's garbage collector. gc.collect() frees memory
# held by objects that are no longer referenced; used below to release an
# old matrix before building a new one.
import gc


def co_occurrence_count(sentences, idx_to_word, word_to_idx, window_size=1):
    """
    Build a word-word co-occurrence matrix based on counts.

    :param sentences: list of sentences, each a list of tokens (strings)
    :param idx_to_word: vocabulary map (index -> word) (dict). Not needed
        for the counting; kept in the signature for compatibility.
    :param word_to_idx: vocabulary map (word -> index) (dict), with
        lowercased words as keys, as built by the tokenization cell
    :param window_size: number of words W taken on EACH side of a word

    :return: symmetric sparse matrix (CSR) of size |V| x |V|
    """
    # Size of the vocabulary, including <PAD> (id 0) and --oov-- (id 1).
    # Their rows and columns simply stay all-zero, but keeping them means
    # row i of the matrix always corresponds to word id i.
    vocab_size = len(word_to_idx)

    # Instead of incrementing a sparse matrix cell by cell (slow in pure
    # Python), collect every co-occurrence as a (row, column) pair in two
    # plain lists. Each pair stands for one count of 1.
    rows, cols = [], []

    # Loop over every sentence (a list of tokens).
    for sentence in sentences:
        # Convert the sentence into word ids once. .lower() matches the
        # lowercasing tokenizer, whose dictionary keys are lowercase. Every
        # word is found, since the vocabulary was built from these same
        # sentences.
        ids = [word_to_idx[w.lower()] for w in sentence]

        # Visit every position p in the sentence; `center` is the id of the
        # word at that position.
        for pos, center in enumerate(ids):
            # First and last positions of the context window, clipped to the
            # sentence boundaries: up to W words before and W words after.
            # For the second word of "the cat is on the table" with W = 2,
            # this gives positions 0..3 -> the, (cat), is, on.
            start = max(0, pos - window_size)
            end = min(len(ids), pos + window_size + 1)  # exclusive end

            # Every position inside the window is a context position.
            for ctx_pos in range(start, end):
                # Skip the center word itself. Comparing POSITIONS (not
                # strings) means a word is never paired with its own
                # occurrence, whatever its spelling.
                if ctx_pos == pos:
                    continue
                # Also skip a nearby repetition of the same word (e.g.
                # "the ... the"), so the diagonal stays zero: a word is
                # not counted as its own context.
                if ids[ctx_pos] == center:
                    continue
                # Record one co-occurrence: row = center word,
                # column = context word. The symmetric pair
                # (context, center) is recorded when the loop reaches the
                # context word's own position, so the matrix is symmetric.
                rows.append(center)
                cols.append(ids[ctx_pos])

    # One count per recorded pair. int32 holds values up to ~2.1 billion,
    # so frequent pairs such as ("of", "the") cannot overflow. (The earlier
    # uint8 type wrapped around silently after 255.)
    data = np.ones(len(rows), dtype=np.int32)

    # Build the matrix in COO format ("coordinate": value, row, column
    # triplets). Duplicate (row, column) pairs are SUMMED when converting
    # to CSR, which turns the individual 1s into co-occurrence counts.
    # CSR ("Compressed Sparse Row") is compact and fast for row slicing,
    # arithmetic and matrix products, which later analyses need.
    matrix = scipy.sparse.coo_matrix((data, (rows, cols)),
                                     shape=(vocab_size, vocab_size)).tocsr()
    return matrix


# Window size W: number of words taken on each side. W = 1 means only
# directly adjacent words count as context (this reproduces the matrix
# built previously). Larger values (e.g. 2-5) capture broader context;
# small windows tend to reflect syntactic similarity, which is the kind
# relevant for POS tagging.
window_size = 1

# Clean RAM before re-running this cell: if a matrix from a previous run
# exists, delete it and force garbage collection, so the old and the new
# matrix are never in memory at the same time.
if 'train_co_occurrence_matrix' in globals():
    del train_co_occurrence_matrix
    gc.collect()

print("Building co-occurrence count matrix... (it may take a while...)")
# Build the matrix from the training + validation sentences, the same
# sentences the vocabulary was built from. list(...) + list(...) joins them
# into one list; the original lists are not modified.
train_co_occurrence_matrix = co_occurrence_count(
    list(train_sentences) + list(validation_sentences),
    idx_to_word, word_to_idx, window_size=window_size)
print("Building completed!")


def evaluate_co_occurrence_matrix(matrix):
    """Consistency checks on the co-occurrence matrix."""
    # Detect whether the matrix is a scipy sparse matrix (true here) or a
    # dense NumPy array, since the checks are written differently.
    is_sparse = scipy.sparse.issparse(matrix)
    if is_sparse:
        print("Detected sparse co-occurrence matrix!")

    # Symmetry: "i occurs near j" implies "j occurs near i", so cell (i, j)
    # must equal cell (j, i).
    print("[Co-occurrence count matrix Evaluation] Symmetry checking...")
    if is_sparse:
        # matrix != matrix.T is a sparse boolean matrix that is True where
        # the two differ; .nnz counts those entries. Zero means symmetric.
        # This avoids building a dense 9,903 x 9,903 array.
        assert (matrix != matrix.transpose()).nnz == 0
    else:
        assert np.equal(matrix, matrix.transpose()).all()

    # Diagonal: a word is never counted as its own context.
    print("[Co-occurrence count matrix Evaluation] Diagonal checking...")
    assert matrix.diagonal().sum() == 0

    # Counts can never be negative (a negative value would reveal an
    # overflow or a construction error).
    print("[Co-occurrence count matrix Evaluation] Sign checking...")
    assert matrix.min() >= 0


print("Evaluating co-occurrence matrix")
evaluate_co_occurrence_matrix(train_co_occurrence_matrix)
print("Evaluation completed!")

# Summary statistics:
# - shape: |V| x |V|;
# - nnz: number of non-zero cells, i.e. distinct (word, context) pairs;
# - density: fraction of cells that are non-zero, showing how sparse the
#   matrix is and why a sparse format is needed;
# - max: the largest single count, i.e. the most frequent adjacent pair.
n_cells = train_co_occurrence_matrix.shape[0] * train_co_occurrence_matrix.shape[1]
print(f"Shape: {train_co_occurrence_matrix.shape}")
print(f"Non-zero entries: {train_co_occurrence_matrix.nnz:,}")
print(f"Density: {train_co_occurrence_matrix.nnz / n_cells:.4%}")
print(f"Largest count: {train_co_occurrence_matrix.max():,}")
```

### Dense Embeddings of Words

Sparse representations such as one-hot vectors or rows of the co-occurrence matrix have one dimension per vocabulary word (|V| = 9,903 here), are mostly zeros, and treat every word as unrelated to every other word. **Dense embeddings** instead map each word to a vector in a **much lower-dimensional space** (typically 50 to 300 dimensions), where every dimension is used by every word.

Dense representations are convenient for machine learning:

- **Fewer parameters:** the downstream model receives small dense inputs instead of |V|-dimensional one-hot vectors, which reduces model size and the risk of overfitting. This matters here, since we have only about 2,000 training sentences.
- **Shared structure:** words no longer occupy separate dimensions. Words used in similar contexts get similar vectors, so what the model learns about one word carries over to similar words, including words that are rare in our training data.
- **Syntactic and semantic regularities:** nouns cluster with nouns, past-tense verbs with past-tense verbs, and related words lie close together. For POS tagging, the syntactic information is especially useful.

Instead of learning embeddings from our small corpus, we use a **pre-trained** model: **GloVe** (Global Vectors), trained on Wikipedia and Gigaword news text with a 400,000-word vocabulary. GloVe is built from global word co-occurrence statistics, the same kind of information as our co-occurrence matrix, but computed on billions of tokens and compressed into dense vectors.

Pre-trained models can be resource-demanding, depending on the embedding dimension and vocabulary size, so we stick to low dimensions (50, 100 or 300) to keep download and loading times short.

We download GloVe with the **[Gensim](https://radimrehurek.com/gensim/)** library, which provides a simple downloader for common pre-trained embedding models.

```python
# ============================================================================
# INSTALL GENSIM
# Gensim is a Python library for topic modelling and word embeddings. Here
# it is used for its downloader module (gensim.downloader), which fetches
# pre-trained embedding models such as GloVe by name and loads them as
# KeyedVectors objects: a word -> vector lookup with similarity functions.
# Gensim is not preinstalled on Colab, so it must be installed first.
# ============================================================================

# The leading "!" tells Jupyter/Colab to run the rest of the line as a
# shell command instead of Python code. "pip install gensim" downloads the
# latest compatible Gensim release from PyPI and installs it into the
# notebook's environment, together with any dependencies it needs.
# If it is already installed, pip only reports "Requirement already
# satisfied" and changes nothing.
!pip install gensim
```

```python
# ============================================================================
# LOAD A PRE-TRAINED WORD EMBEDDING MODEL (GloVe)
# Downloads a pre-trained embedding model through Gensim's downloader and
# loads it into memory as a KeyedVectors object: a lookup table from each
# word in the model's vocabulary to its dense vector. These vectors will
# fill the rows of the embedding matrix used by the tagging models.
# ============================================================================

#So, basically we are trying to embed the sentences above, what we can do is use glove embedding to achieve it
# (More precisely: GloVe gives one vector per WORD, not per sentence. Each
# sentence is represented as the sequence of its words' vectors, which the
# recurrent models then read in order.)

# The Gensim package itself. Not used directly below, but importing it
# confirms the installation from the previous cell worked.
import gensim

# Gensim's downloader API, under the short alias `gloader`. It knows a
# catalogue of pre-trained models and corpora by name, downloads them on
# first use to a local cache folder (~/gensim-data), and loads them.
import gensim.downloader as gloader


def load_embedding_model(model_type, embedding_dimension=50):
    """
    Loads a pre-trained word embedding model via gensim library.

    :param model_type: name of the word embedding model to load.
    :param embedding_dimension: size of the embedding space to consider

    :return
        - pre-trained word embedding model (gensim KeyedVectors object)
    """

    # Will hold the name of the model in Gensim's catalogue.
    download_path = ""

    # Find the correct embedding model name
    # .strip() removes surrounding spaces and .lower() makes the comparison
    # case-insensitive, so "Glove", " glove " and "GLOVE" all match.
    if model_type.strip().lower() == 'word2vec':
        # Word2Vec trained on Google News (about 100 billion words). It is
        # only available with 300 dimensions, so embedding_dimension is
        # ignored for this option. It is large (about 1.6 GB download,
        # 3 million words and phrases) and case-sensitive, unlike our
        # lowercased vocabulary.
        download_path = "word2vec-google-news-300"

    elif model_type.strip().lower() == 'glove':
        # GloVe trained on Wikipedia 2014 + Gigaword 5 news text: 400,000
        # lowercased words. The catalogue name includes the dimension, e.g.
        # "glove-wiki-gigaword-300"; .format inserts the requested size.
        download_path = "glove-wiki-gigaword-{}".format(embedding_dimension)

    else:
        # Any other model name is rejected with an explanatory message.
        # (ValueError would be the more conventional exception type for an
        # invalid argument value; AttributeError still stops execution.)
        raise AttributeError("Unsupported embedding model type! Available ones: word2vec, glove")

    # Check download
    try:
        # Download the model if it is not cached yet, then load it into RAM
        # as a KeyedVectors object. On later runs it is read from the local
        # cache, which is much faster.
        emb_model = gloader.load(download_path)
    except ValueError as e:
        # Gensim raises ValueError for a name it does not know, e.g.
        # "glove-wiki-gigaword-150", since GloVe exists only in the four
        # sizes listed below. Print the valid options, then re-raise the
        # original error so the cell still stops.
        print("Invalid embedding model name! Check the embedding dimension:")
        print("Word2Vec: 300")
        print("Glove: 50, 100, 200, 300")
        raise e

    # Return the loaded model. Useful operations on it:
    #   emb_model['board']                      -> its 300-dim vector
    #   'board' in emb_model                    -> whether it is in GloVe
    #   emb_model.most_similar('board')         -> nearest words
    #   emb_model.key_to_index / index_to_key   -> its own vocabulary
    return emb_model


# Modify these variables as you wish!
# Glove -> 50, 100, 200, 300
# Word2Vec -> 300
# Model family to load; matched case-insensitively above.
embedding_model_type = "Glove"

# Size of each word vector. 300 is the largest GloVe version: the most
# expressive, but also the slowest to download (about 376 MB) and the
# largest in memory (400,000 x 300 floats, roughly 0.5 GB). It also sets
# the output size of the Embedding layer, and therefore the input size of
# the recurrent layers.
embedding_dimension = 300

# Download (first time only) and load the model. The first run can take a
# few minutes, while Gensim shows a progress bar.
embedding_model = load_embedding_model(embedding_model_type, embedding_dimension)
```

### Out of vocabulary (OOV) words

Before evaluating pre-trained dense word embeddings, it is good practice to check if the model is consistent with our dataset. To do so, we check the number of out-of-vocabulary (OOV) terms.

If the OOV amount is negligible, we can just keep going. On the other hand, we might want to handle OOV terms by assigning them a specific word vector.

**Which one?** One common practice is to assign a random vector, since the embedding model will be part of a deep learning model and, thus, word vectors might be trained during the learning process. Even if that is the case, we can assign an embedding that is more meaningful rather than a random one: for example, we can identify the word embedding of an OOV term as the mean of its neighbour word embeddings.


```python
# ============================================================================
# CHECK OUT-OF-VOCABULARY (OOV) TERMS WITH RESPECT TO GloVe
# Compares our vocabulary (built from the training + validation sentences)
# with GloVe's vocabulary, and lists the words GloVe does not know. These
# words will need a vector of our own (random, or the mean of their
# neighbours) when the embedding matrix is built.
# Note: "OOV" here means "in our vocabulary but missing from GloVe", not
# the tokenizer's --oov-- token (words missing from OUR vocabulary).
# ============================================================================

# Function definition

def check_OOV_terms(embedding_model, word_listing):
    """
    Checks differences between pre-trained embedding model vocabulary
    and dataset specific vocabulary in order to highlight out-of-vocabulary terms.

    :param embedding_model: pre-trained word embedding model (gensim KeyedVectors)
    :param word_listing: dataset specific vocabulary (list)

    :return
        - alphabetically sorted list of OOV terms
    """
    # embedding_model.key_to_index is GloVe's word -> row dictionary
    # (Gensim 4; older versions called it .vocab). Converting its keys and
    # our word list to sets allows a set difference: the words in our
    # vocabulary that are NOT in GloVe. Set membership tests are fast, so
    # this takes a fraction of a second even with 400,000 GloVe words.
    # A set has no order, so the result is sorted to make the printout
    # reproducible from run to run.
    return sorted(set(word_listing) - set(embedding_model.key_to_index.keys()))


# The vocabulary was already built in the tokenization cell from the
# training + validation sentences, so it is reused here instead of fitting
# the tokenizers again. The "train_" names are kept for the cells that
# follow; note that they cover training AND validation words.
train_word_listing = word_listing     # 9,901 distinct lowercased words
train_word_to_idx = word_to_idx       # word -> id (with <PAD> and --oov--)
train_idx_to_word = idx_to_word       # id -> word

# Alternative (kept commented out): a vocabulary from the training
# sentences only. words_tokenizer returns 7 values, so all 7 must be
# unpacked; this would also overwrite the tokenizers and dictionaries.
#word_tokenizer, tag_tokenizer, word_listing, word_to_idx, idx_to_word, tag_to_idx, idx_to_tag = words_tokenizer(list(train_sentences), list(train_tags))

# Words of our vocabulary that have no GloVe vector.
train_oov_terms = check_OOV_terms(embedding_model, train_word_listing)

# ---------------------------------------------------------------------------
# How much does it matter? Two measures:
# - by word TYPE: share of distinct vocabulary words missing from GloVe;
# - by TOKEN: share of word OCCURRENCES in the text that are missing.
# The token share is the more relevant one: a few hundred rare missing
# words matter much less than one frequent missing token.
# ---------------------------------------------------------------------------

# word_tokenizer.word_counts holds each (lowercased) word's number of
# occurrences in the training + validation sentences.
total_tokens = sum(word_tokenizer.word_counts.values())
oov_tokens = sum(word_tokenizer.word_counts[w] for w in train_oov_terms)

# Percentages must be multiplied by 100 (the original commented-out print
# showed the fraction, e.g. 0.05, followed by a % sign).
print("Vocabulary size: {}".format(len(train_word_listing)))
print("OOV terms (types): {0} ({1:.2f}% of the vocabulary)".format(
    len(train_oov_terms), 100 * len(train_oov_terms) / len(train_word_listing)))
print("OOV occurrences (tokens): {0} ({1:.2f}% of all tokens)".format(
    oov_tokens, 100 * oov_tokens / total_tokens))

# The 20 most frequent OOV terms, with their counts: these are the ones
# whose vectors matter most for the model.
print("\nMost frequent OOV terms:")
most_frequent_oov = sorted(train_oov_terms,
                           key=lambda w: word_tokenizer.word_counts[w],
                           reverse=True)[:20]
print([(w, word_tokenizer.word_counts[w]) for w in most_frequent_oov])

# Print all OOV terms, 10 per line. Using range(0, len, 10) with slicing
# also prints the last, incomplete group. (The original loop,
# range(len // 10), silently skipped up to 9 terms at the end.)
print("\nAll OOV terms:")
for i in range(0, len(train_oov_terms), 10):
    print(train_oov_terms[i:i + 10])
```

### Why a pie chart for GloVe coverage?

**The type of data**

The variable shown is **categorical (nominal) with two categories**: every word of our vocabulary is either *found in GloVe* or *OOV* (missing from GloVe). Each word belongs to exactly one category, so the two shares together always make up **100 %** of the vocabulary.

The question we ask is about **composition**: *what fraction of the whole vocabulary does GloVe cover, and what fraction is missing?* This is a **part-to-whole** comparison.

**Why a pie chart fits**

- A pie chart represents a **whole divided into parts**: the full circle is the complete vocabulary and each slice is one category's share. This matches the part-to-whole question directly.
- With **only two categories**, the chart is immediately readable: one glance shows that the covered part dominates and how large the missing part is.
- The `autopct` labels print the exact percentage inside each slice, so the chart shows both the visual proportion and the precise value.

**When a pie chart would not fit**

- **Many categories:** with more than about 4–5 slices, the angles become hard to compare and the labels overlap. This is why the tag distribution (45 tags) was shown with a bar chart instead.
- **Comparing values that do not form a whole:** e.g. the sentence counts of the three splits could be compared with a pie chart, but the bar chart used earlier made it easier to compare the sizes against each other.
- **Very small slices:** people judge angles and areas less accurately than lengths. If the OOV share is only a few percent, its slice is thin and the exact number (from the label or the printed percentage) matters more than the picture.

**Alternatives**

| Visualization | Suited to | Here |
|---------------|-----------|------|
| **Pie chart** | Part-to-whole with 2–4 categories | Clear overview of coverage vs. OOV |
| Stacked horizontal bar | Part-to-whole, especially with small shares or several wholes side by side | Better for comparing coverage by **type** and by **token** in one figure |
| Bar chart | Comparing counts or values of several categories | Better for listing the most frequent OOV words and their counts |
| Plain text / table | One or two key numbers | The two percentages alone would also be sufficient |

**What the chart does not show**

The percentages are computed over **word types** (each distinct word counts once). Most OOV words are rare, so their share of actual **word occurrences** in the text is usually much smaller. The token-level share, printed by the previous cell, better reflects how often the model will meet a word without a GloVe vector.

## Building the Embedding Matrix

### Why an embedding matrix is needed

After tokenization, every sentence is a sequence of integer ids, and each id is only a label for a word. The models need a **meaningful vector for every id**. The **embedding matrix** provides these vectors: it has one row per vocabulary id (|V| = 9,903 rows) and one column per embedding dimension (*d* = 300 for the GloVe model we loaded).

Row *i* holds the vector of the word with id *i*. This matrix is loaded into the **Embedding layer** of each model. For every id in an input sentence, the layer looks up the corresponding row, so a sentence of *n* ids becomes an *n* × *d* sequence of vectors, which the recurrent layers (LSTM, GRU, ...) read in order.

Initialising the matrix with **pre-trained GloVe vectors**, instead of random values, gives the models knowledge learned from billions of words of text. This matters for our small dataset (about 2,000 training sentences): most words appear too rarely for the model to learn good vectors from scratch.

### Why OOV words need special handling

GloVe does not contain every word of our vocabulary. Penn Treebank-specific tokens (e.g. bracket tokens such as `-lrb-`), some number formats, hyphenated compounds and rare names are missing. These **OOV terms** (in our vocabulary but not in GloVe) still need a row in the matrix, because every id that can appear in the input must map to a vector. The way we fill these rows affects how well the model handles those words:

- A **zero vector** would give the model no information and make all OOV words identical.
- A **random vector** gives each OOV word a distinct representation, which training can then adjust, but only for words that appear in the training sentences. Words that occur only in the validation sentences never receive a gradient update and keep their random, meaningless vector.
- A vector derived from the word's **context** places the OOV word near the words it is used with, so it starts with a meaningful representation even if training never updates it.

### Special tokens

- **`<PAD>` (id 0)** gets a **zero vector**. Padded positions carry no information and are masked out by the model.
- **`--oov--` (id 1)** represents words unseen in our vocabulary (e.g. new test words). It has no context of its own, so it gets a random vector.

### Strategy for each word *w*

1. **If *w* is in the GloVe vocabulary**, assign its GloVe vector.
2. **Otherwise**, *w* is an OOV term, and one of two strategies applies:
   - **Random strategy:** assign a random vector. Its values are drawn from a range that matches the GloVe vectors (e.g. based on their mean and standard deviation), so the new vectors are on the same scale as the pre-trained ones and do not dominate or vanish in the model's computations.
   - **Neighbour strategy:**
     1. Take the row of *w* in the **co-occurrence matrix** (window size 1), which lists the words that appear directly before or after *w*.
     2. For each neighbour with a count ≥ 1 that **has a GloVe vector** (neighbours that are OOV themselves are skipped), retrieve that vector.
     3. Compute the **weighted average** of these vectors, using the co-occurrence counts as weights, so frequent neighbours contribute more, and assign it to *w*.
     4. **If *w* has no usable neighbours** (e.g. it occurs only in the validation sentences, which were not used to build the co-occurrence matrix, or all its neighbours are OOV), fall back to the **random strategy**.

### Why the neighbour strategy

The neighbour strategy follows the **distributional hypothesis** that also underlies GloVe: *words that appear in similar contexts have similar meanings and grammatical roles*. An OOV word surrounded by nouns and determiners is likely to behave like them, so averaging its neighbours' vectors places it in a sensible region of the embedding space, close to the words it is used with. This gives the model a useful starting point instead of arbitrary noise, and it reuses the co-occurrence matrix built earlier.

One limitation to keep in mind: raw co-occurrence counts are dominated by very frequent words (e.g. *the*, *of*, `,`), so the averages of different OOV words tend to be similar. Down-weighting or excluding the most frequent neighbours (e.g. with PPMI weights) can make the resulting vectors more distinctive.

### Strategies tested

 **neighbour strategy** as our final implementation, shown below.

```python
# ============================================================================
# BUILD THE EMBEDDING MATRIX
# Creates a |V| x d matrix (|V| = vocabulary size including <PAD> and
# --oov--, d = GloVe dimension) where row i is the vector of the word with
# id i. It will initialise the Embedding layer of every model, which looks
# up row i for each id i in the input sentences.
#   - words in GloVe            -> their GloVe vector
#   - GloVe-OOV words           -> depends on the chosen heuristic
#                                  ('none', 'random' or 'neighbour')
#   - <PAD> (id 0)              -> zero vector (padding carries no information)
#   - --oov-- (id 1)            -> random vector (no context of its own)
# ============================================================================

import scipy.sparse

# Fixed seed for NumPy's random generator, so the random vectors (and thus
# the models' starting point) are the same on every run.
rng = np.random.default_rng(seed=42)


def random_heuristic(embedding_dimension, mean, std):
    """Random vector on the same scale as the GloVe vectors: each dimension
    is drawn from a normal distribution with that dimension's mean and
    standard deviation over the known GloVe vectors. (A uniform draw between
    the global min and max would give vectors much larger than typical GloVe
    vectors, because most GloVe values lie close to 0.)"""
    return rng.normal(loc=mean, scale=std, size=embedding_dimension)


def zero_heuristic(embedding_dimension):
    """All-zero vector: gives the model no information about the word."""
    return np.zeros(embedding_dimension, dtype=np.float32)


def neighbour_heuristic(word_id, co_occurrence_matrix, embedding_matrix, known_mask):
    """Weighted average of the GloVe vectors of the word's neighbours.

    Reads row `word_id` of the co-occurrence matrix (window size 1: words
    appearing directly before or after the word) and averages the vectors
    of the neighbours that have a GloVe vector, weighted by how often each
    neighbour co-occurs with the word. Returns None if there is no usable
    neighbour, so the caller can fall back to a random vector."""
    # Extract the sparse row: .indices are the column ids with a non-zero
    # count (the neighbours) and .data their counts. Only these few entries
    # are read, instead of looping over all |V| columns.
    row = co_occurrence_matrix.getrow(word_id)
    neighbour_ids, counts = row.indices, row.data

    # Keep only neighbours that have a GloVe vector. Neighbours that are
    # OOV themselves have no reliable vector yet and would pull the average
    # towards zero.
    keep = known_mask[neighbour_ids]
    neighbour_ids, counts = neighbour_ids[keep], counts[keep]

    # No usable neighbour (e.g. the word occurs only in validation
    # sentences, which were not used for the co-occurrence matrix, or all
    # its neighbours are OOV): signal the caller to use the fallback.
    if counts.sum() == 0:
        return None

    # Weighted average: counts (shape k) times the k neighbour vectors
    # (shape k x d) gives the weighted sum (shape d); dividing by the total
    # weight turns it into an average on the same scale as GloVe vectors.
    return counts @ embedding_matrix[neighbour_ids] / counts.sum()


def build_embedding_matrix(embedding_model, embedding_dimension, word_to_idx,
                           oov_terms, oov_terms_heuristic='none',
                           co_occurrence_matrix=None):
    """
    Builds the embedding matrix of the dataset given a pre-trained model.

    :param embedding_model: pre-trained word embedding model (gensim KeyedVectors)
    :param embedding_dimension: size of the vectors (must match the model)
    :param word_to_idx: vocabulary map (word -> index), with <PAD> at 0 and --oov-- at 1
    :param oov_terms: words of our vocabulary missing from the model
    :param oov_terms_heuristic: 'none', 'random' or 'neighbour'
    :param co_occurrence_matrix: |V| x |V| sparse matrix, required for 'neighbour'

    :return: embedding matrix of shape |V| x d (float32)
    """
    # The requested dimension must match the loaded model's vectors.
    assert embedding_dimension == embedding_model.vector_size

    # Validate the heuristic name before doing any work.
    assert oov_terms_heuristic in ('none', 'random', 'neighbour')
    if oov_terms_heuristic == 'neighbour':
        assert co_occurrence_matrix is not None, "the 'neighbour' heuristic needs a co-occurrence matrix"

    # A set gives constant-time membership tests; checking a word against a
    # list of ~1,000 OOV terms 9,903 times would be needlessly slow.
    oov_set = set(oov_terms)

    # Special tokens: they are not real words and are never in GloVe.
    special_tokens = {'<PAD>', '--oov--'}

    vocab_size = len(word_to_idx)   # 9,903: one row per id, no extra row

    # Start from all zeros. float32 is the type Keras uses for weights and
    # halves the memory of NumPy's default float64.
    embedding_matrix = np.zeros((vocab_size, embedding_dimension), dtype=np.float32)

    # known_mask[i] is True if row i was filled from GloVe. It identifies
    # usable neighbours and the rows used for the random-vector statistics.
    known_mask = np.zeros(vocab_size, dtype=bool)

    # --- Step 1: copy the GloVe vector of every word GloVe knows.
    for word, idx in word_to_idx.items():
        if word not in special_tokens and word not in oov_set:
            embedding_matrix[idx] = embedding_model[word]
            known_mask[idx] = True

    # Per-dimension mean and standard deviation of the GloVe vectors of our
    # known words, used to draw random vectors on the same scale.
    mean = embedding_matrix[known_mask].mean(axis=0)
    std = embedding_matrix[known_mask].std(axis=0)

    # --- Step 2: fill the rows of the GloVe-OOV words.
    n_neighbour, n_random = 0, 0          # counters for the report below
    for word in oov_terms:
        idx = word_to_idx[word]
        if oov_terms_heuristic == 'none':
            embedding_matrix[idx] = zero_heuristic(embedding_dimension)
        elif oov_terms_heuristic == 'random':
            embedding_matrix[idx] = random_heuristic(embedding_dimension, mean, std)
            n_random += 1
        else:  # 'neighbour'
            vector = neighbour_heuristic(idx, co_occurrence_matrix,
                                         embedding_matrix, known_mask)
            if vector is not None:
                embedding_matrix[idx] = vector
                n_neighbour += 1
            else:
                # Fallback when the word has no usable neighbour.
                embedding_matrix[idx] = random_heuristic(embedding_dimension, mean, std)
                n_random += 1

    # --- Step 3: special tokens.
    # <PAD> stays a zero vector (it is masked out by the model anyway).
    # --oov-- has no context of its own, so it gets a random vector, except
    # with 'none', where every non-GloVe row stays zero.
    if oov_terms_heuristic != 'none':
        embedding_matrix[word_to_idx['--oov--']] = random_heuristic(embedding_dimension, mean, std)

    # Report how the OOV rows were filled.
    print(f"Heuristic: {oov_terms_heuristic}")
    print(f"  words from GloVe:            {known_mask.sum()}")
    print(f"  OOV words, neighbour average: {n_neighbour}")
    print(f"  OOV words, random vector:     {n_random}")

    return embedding_matrix


# Reuse the vocabulary and OOV list from the previous cells (no re-fitting
# of the tokenizers). The co-occurrence matrix built earlier from the
# training sentences with window size 1 provides the neighbours.
train_embedding_matrix = build_embedding_matrix(
    embedding_model, embedding_dimension, train_word_to_idx, train_oov_terms,
    oov_terms_heuristic='neighbour',
    co_occurrence_matrix=train_co_occurrence_matrix)

# Expected shape: (9903, 300) -> one row per vocabulary id, one column per
# GloVe dimension. It must match the Embedding layer's (input_dim, output_dim).
print("Embedding matrix shape: {}".format(train_embedding_matrix.shape))

# Sanity checks:
# - <PAD> is the zero vector;
# - no other row is entirely zero, i.e. every word received a vector.
assert not train_embedding_matrix[0].any()
assert np.all(np.abs(train_embedding_matrix[1:]).sum(axis=1) > 0)
print("Embedding matrix checks passed!")
```

### Pad sequences

The sentences in the corpus have different lengths. A recurrent model can process a single sentence of any length, but training is done on **batches**, and a batch is one tensor of shape (batch size, sequence length). All sequences must therefore be brought to the **same length** before they are fed to the model.

We do this by:

- **padding** shorter sequences with the `<PAD>` id (0) up to a fixed length, and
- **truncating** sequences that are longer than that length.

Padding is added **after** the words (`padding='post'`), so every sentence starts at position 0. Padded positions carry no information: the Embedding layer masks them (`mask_zero=True`), and they are excluded from the loss and the evaluation metrics. The tag sequences are padded in exactly the same way, so that each word stays aligned with its tag.

The fixed length is a **hyperparameter**: a longer length keeps more of the long sentences but makes every batch larger and slower to process. To choose it, we looked at the sentence-length distribution of each split:

| Split | Longest sentence | 99th percentile | Sentences > 100 tokens | Tokens cut |
|-------|------------------|-----------------|------------------------|------------|
| Train | 250 | 57 | 3 | 175 |
| Validation | 88 | 61 | 0 | 0 |
| Test | 80 | 52 | 0 | 0 |

99 % of the sentences have at most about 60 tokens. The 250-token maximum is an outlier, most likely several sentences merged by the period-based sentence splitting. We choose a length of **100 tokens**:

- **no validation or test token is lost**, so the evaluation covers the complete data;
- only **3 training sentences** are truncated, losing 175 tokens (well under 1 % of the training tokens);
- the length leaves a margin above the longest validation and test sentences, while avoiding the cost of padding every sequence to 250.

```python
# ============================================================================
# SENTENCE-LENGTH STATISTICS BEFORE PADDING
# Why this cell is needed:
# Before padding, we must choose a fixed sequence length (here 100). This
# length is a hyperparameter with a trade-off:
#   - too short: long sentences are truncated, and the cut words receive
#     no prediction. In the validation and test sets this would silently
#     remove tokens from the evaluation, making the scores unreliable.
#   - too long: every sequence is padded to that length, so most positions
#     are padding and training becomes slower for no benefit.
# This cell measures, for each split, the longest sentence, the typical
# upper range of lengths (99th percentile), and how many sentences and
# tokens a length of 100 would cut. It shows whether 100 is a safe choice.
# ============================================================================

# Sentence-length statistics per split, and how many sentences exceed 100 tokens.
# Loop over the three splits. Each item of the list is a (name, sentences)
# pair, unpacked into `name` (used in the printout) and `sents` (a list of
# token lists).
for name, sents in [('train', train_sentences), ('validation', validation_sentences), ('test', test_sentences)]:

    # Length (number of tokens) of every sentence in the split, as a NumPy
    # array so that vectorised operations (max, percentile, comparisons)
    # can be applied to all lengths at once.
    lengths = np.array([len(s) for s in sents])

    # Print one summary line per split. The f-string is split over two
    # lines; Python joins adjacent string literals automatically.
    # - lengths.max(): the longest sentence. It shows the largest length
    #   that would be needed to avoid any truncation at all.
    # - np.percentile(lengths, 99): the length below which 99 % of the
    #   sentences fall. Unlike the maximum, it is not driven by a single
    #   outlier, so it shows what length the typical sentence needs.
    #   ':.0f' prints it as a whole number.
    # - (lengths > 100).sum(): lengths > 100 is a boolean array (True for
    #   each sentence longer than 100 tokens); summing it counts the Trues,
    #   i.e. the sentences that would be truncated.
    # - (lengths[lengths > 100] - 100).sum(): selects only the lengths above
    #   100 (boolean indexing), subtracts 100 from each to get how many
    #   tokens each such sentence would lose, and adds them up: the total
    #   number of tokens removed by truncation.
    print(f"{name}: max {lengths.max()}, 99th percentile {np.percentile(lengths, 99):.0f}, "
          f"> 100 tokens: {(lengths > 100).sum()} sentences ({(lengths[lengths > 100] - 100).sum()} tokens cut)")
```

```python
# ============================================================================
# PAD (AND TRUNCATE) ALL SEQUENCES TO A FIXED LENGTH
# Training works on batches, and a batch is a single tensor of shape
# (batch size, sequence length), so every sequence must have the same
# length. Each encoded sentence and each encoded tag sequence is:
#   - padded with 0 (the <PAD> id) at the END if shorter than
#     MAX_SEQ_LENGTH, and
#   - cut at the END if longer.
# X and y are padded with exactly the same settings, so every word id
# stays aligned with its tag id, position by position, and every padded
# word position (id 0) matches a padded tag position (class 0).
# ============================================================================

# In Keras 3, pad_sequences is in keras.utils (the old paths
# keras.preprocessing.sequence and the separate keras_preprocessing
# package are no longer available).
from keras.utils import pad_sequences

# Length of the longest training sentence, for reference (250 tokens).
# max(..., key=len) returns the longest sentence; len() gives its length.
print('Max Sequence Length: ', len(max(train_sentences, key=len)))

# Fixed length chosen from the statistics of the previous cell: longer
# than every validation and test sentence (88 and 80 tokens), so no
# evaluation token is lost, while only 3 training sentences are cut.
MAX_SEQ_LENGTH = 100
print('Defined Sequence Length: ', MAX_SEQ_LENGTH)

# pad_sequences takes a list of variable-length lists and returns a NumPy
# array of shape (number of sequences, MAX_SEQ_LENGTH), dtype int32.
# - maxlen=MAX_SEQ_LENGTH: target length for every sequence.
# - padding='post': zeros are added AFTER the words, so every sentence
#   starts at position 0. Keras' LSTM/GRU layers can then use the fast
#   cuDNN implementation on GPU, which requires right-padded masks.
# - truncating='post': words beyond position 100 are removed from the
#   END, so the beginning of the sentence is kept. (Keras' default,
#   'pre', would remove words from the beginning.)
# - value=0 (the default): the padding id, i.e. <PAD> in both
#   vocabularies.
X_train_padded = pad_sequences(X_train_encoded, maxlen=MAX_SEQ_LENGTH, padding='post', truncating='post')
X_validation_padded = pad_sequences(X_validation_encoded, maxlen=MAX_SEQ_LENGTH, padding='post', truncating='post')
X_test_padded = pad_sequences(X_test_encoded, maxlen=MAX_SEQ_LENGTH, padding='post', truncating='post')

# Same padding for the tag sequences: padded positions get tag class 0
# (<PAD>), which is later excluded from the loss and the metrics.
y_train_padded = pad_sequences(y_train_encoded, maxlen=MAX_SEQ_LENGTH, padding='post', truncating='post')
y_validation_padded = pad_sequences(y_validation_encoded, maxlen=MAX_SEQ_LENGTH, padding='post', truncating='post')
y_test_padded = pad_sequences(y_test_encoded, maxlen=MAX_SEQ_LENGTH, padding='post', truncating='post')

# Sanity checks:
# - every array now has exactly MAX_SEQ_LENGTH columns;
# - X and y have zeros in exactly the same positions, i.e. padding is
#   aligned between words and tags.
for X_pad, y_pad in [(X_train_padded, y_train_padded),
                     (X_validation_padded, y_validation_padded),
                     (X_test_padded, y_test_padded)]:
    assert X_pad.shape == y_pad.shape and X_pad.shape[1] == MAX_SEQ_LENGTH
    assert np.array_equal(X_pad == 0, y_pad == 0)

# Shapes: (number of sentences, 100) for each split.
print('Train shapes:     ', X_train_padded.shape, y_train_padded.shape)
print('Validation shapes:', X_validation_padded.shape, y_validation_padded.shape)
print('Test shapes:      ', X_test_padded.shape, y_test_padded.shape)

# print the first sequence
# The 18 word ids of the first sentence, followed by 82 zeros of padding.
print(X_train_padded[0], "\n"*3)
# The 18 tag ids of the same sentence, followed by 82 zeros.
print(y_train_padded[0])
```

### Visualization of Tags

We inspect the tag vocabulary built by the tag tokenizer (`tag_to_idx`) to see which ids correspond to which tags, and in particular to identify the **punctuation tags**: `,` `.` `:` ` `` ` `''` `-LRB-` `-RRB-`, and possibly the symbol tags `$` and `#`, depending on the convention chosen.

Punctuation tokens are kept in the data: they remain part of the input sentences, where they provide useful context for the neighbouring words, and the models still learn to predict their tags. However, punctuation tags are **excluded from the evaluation**. Each punctuation token has its own dedicated tag, so predicting it is trivial, and since punctuation is very frequent, including it would inflate the scores and hide the differences between models on the real words.

We therefore build the list of tag ids **without the punctuation tags** (and without the padding class 0). These are the classes on which the macro F1 score and the other metrics will be computed.

```python
# ============================================================================
# TAG VOCABULARY AND EVALUATION LABELS (WITHOUT PUNCTUATION)
# Shows the tag -> id mapping, then builds the list of tag ids on which
# the models will be evaluated. Padding and punctuation classes are left
# out:
#   - <PAD> (id 0) marks padded positions, not real tokens.
#   - Punctuation tags are trivial to predict (each punctuation mark has
#     its own tag) and very frequent, so including them would inflate the
#     scores and hide differences between models on the real words.
# Punctuation tokens are NOT removed from the data: they stay in the input
# as context, and the models still predict their tags. They are only
# excluded from the metrics.
# ============================================================================

# tag_to_idx was built in the tokenization cell: {'<PAD>': 0, 'NN': 1, ...}.
# It is reused as it is (reassigning it from tag_tokenizer.word_index would
# drop the <PAD> entry). Printed to see which id corresponds to which tag;
# ids follow tag frequency in the training data.
print(tag_to_idx)

# Tags excluded from the evaluation.
# Penn Treebank punctuation tags:
#   ','      comma
#   '.'      sentence-final punctuation (also used for '?' and '!')
#   ':'      colon, semicolon, dash, ellipsis
#   '``'     opening quotation mark
#   "''"     closing quotation mark
#   '-LRB-'  left bracket  "(" (Left Round Bracket)
#   '-RRB-'  right bracket ")" (Right Round Bracket)
# Symbol tags, treated as punctuation here because they are just as
# trivial to predict:
#   '$'      currency sign
#   '#'      pound sign
# Plus the padding class '<PAD>', which is not a real tag.
punctuations = [',', '.', ':', '``', "''", '-LRB-', '-RRB-', '$', '#', '<PAD>']

# Sanity check: every excluded name must be an actual entry of the tag
# vocabulary. A misspelled or non-existent tag (e.g. '?', or the lowercase
# '-lrb-') would otherwise be silently ignored, and that class would stay
# in the evaluation.
missing = [p for p in punctuations if p not in tag_to_idx]
assert not missing, f"Not in the tag vocabulary: {missing}"

# Let's list the labels of non punctuations symbols
# Ids of all remaining tags: the classes used for the evaluation metrics
# (e.g. the `labels` argument of f1_score, which restricts macro-averaging
# to these classes).
labels = [idx for tag, idx in tag_to_idx.items() if tag not in punctuations]

# Same selection by name, for readable output and for labelling plots.
label_names = [tag for tag, idx in tag_to_idx.items() if tag not in punctuations]

print(label_names)
print(labels)
# Expected: 45 tags - 9 punctuation/symbol tags = 36 evaluation labels.
print('Labels size: {}'.format(len(labels)))
```

    Shape of X: (#samples, #timesteps, #features)

- **#samples:** number of sentences in the split;

## RNN Models

https://en.wikipedia.org/wiki/Recurrent_neural_network

### What is a Recurrent Neural Network (RNN)?

A **Recurrent Neural Network** is a neural network designed for **sequential data**, such as the words of a sentence. A standard feed-forward network processes each input independently. An RNN instead reads the sequence **one element at a time** and keeps a **hidden state**, a vector that summarises what it has seen so far. At each step, the hidden state is updated from the current input and the previous hidden state, so information from earlier words carries over to later ones.

This is what POS tagging needs: the tag of a word depends on its context. In *"they **board** the plane"*, knowing that *"board"* follows the pronoun *"they"* makes the verb reading (`VBP`) much more likely than the noun reading (`NN`).

### Structure of an RNN

At every timestep *t*, the RNN cell receives two inputs, the current word vector **xₜ** and the previous hidden state **hₜ₋₁**, and computes the new hidden state:

    hₜ = tanh( Wₓ · xₜ + Wₕ · hₜ₋₁ + b )

- **Wₓ** transforms the current input;
- **Wₕ** transforms the previous hidden state (the "memory" of the preceding words);
- **b** is a bias, and **tanh** keeps the values between −1 and 1;
- **h₀**, before the first word, is a vector of zeros.

For tagging, an output layer turns each hidden state into a probability distribution over the tags:

    ŷₜ = softmax( Wᵧ · hₜ + bᵧ )

The **same weights** (Wₓ, Wₕ, Wᵧ and the biases) are used at every timestep. The network does not learn separate parameters for position 1, position 2, and so on; it applies one cell repeatedly. This is why an RNN can process sequences of any length, and why the number of parameters does not grow with the sentence length.

Drawing one copy of the cell per timestep ("unrolling" the network over the sentence) makes the structure visible:

          ŷ₁           ŷ₂           ŷ₃               tags (softmax)
           ↑            ↑            ↑
    h₀ →  [RNN]  h₁ →  [RNN]  h₂ →  [RNN]  h₃ → ...   hidden states
           ↑            ↑            ↑
           x₁           x₂           x₃               word vectors (embedding)
         "they"      "board"       "the"

Every box is the same cell with the same weights. The arrows between the boxes carry the hidden state from one word to the next.

### RNN architecture for POS tagging

POS tagging is a **many-to-many** task with aligned sequences: one input word and one output tag at every timestep. Our models follow this pipeline:

| Layer | Role | Output shape |
|-------|------|--------------|
| Input | Padded word ids | (batch, 100) |
| **Embedding** | Replaces each id with its word vector; masks padding (`mask_zero=True`) | (batch, 100, 300) |
| **Recurrent layer** | Reads the vectors in order and outputs a hidden state at **every** timestep (`return_sequences=True`) | (batch, 100, units) |
| **Dense + softmax** (applied at each timestep) | Converts each hidden state into probabilities over the 46 tag classes | (batch, 100, 46) |

The model is trained with the **categorical cross-entropy** loss between the predicted distributions and the one-hot encoded tags. Padded positions are ignored thanks to the mask.

### Limitations of the vanilla RNN, and its variants

- **Vanishing gradients:** during training, the error signal is propagated backwards through all the timesteps, being multiplied at every step by the same weights. Over many steps it tends to shrink towards zero (or, more rarely, explode), so a simple RNN struggles to learn dependencies between words that are far apart.
- **LSTM** (Long Short-Term Memory) and **GRU** (Gated Recurrent Unit) cells address this with **gates**: small learned components that decide what to keep, forget and output at each step. They are much better at carrying information across long distances, which is why they are used in the baseline and later models.
- **Only past context:** a simple (unidirectional) RNN predicts the tag of a word using only the words **before** it. A **bidirectional** RNN runs a second RNN from the end of the sentence backwards and combines both, so every prediction uses the context on both sides. This is important for tagging: in *"the **board** met"*, the following word *"met"* confirms that *"board"* is a noun.

### Embeddings: frozen or trainable

The weights of the Embedding layer can be treated in two ways:

- **frozen** (`trainable=False`): the vectors stay fixed during training;
- **trainable** (`trainable=True`): the vectors are updated like any other weight of the network, at every training step.

The embedding matrix has 9,903 × 300 ≈ 3 million values, so making it trainable adds far more parameters than the recurrent layer itself, which increases the risk of overfitting with our roughly 2,000 training sentences. Also, only the vectors of words that appear in the training sentences are updated: words seen only in the validation or test sets keep their initial vectors.

## Baseline Model (Bidirectional LSTM)- Model-1

To solve the vanishing gradients problem, many attempts have been made to tweak the vanilla RNNs such that the gradients don’t die when sequences get long. The most popular and successful of these attempts has been the long, short-term memory network, or the LSTM. LSTMs have proven to be so effective that they have almost replaced vanilla RNNs.

Thus, one of the fundamental differences between an RNN and an LSTM is that an LSTM has an explicit memory unit which stores information relevant for learning some task. In the standard RNN, the only way the network remembers past information is through updating the hidden states over time, but it does not have an explicit memory to store information.

On the other hand, in LSTMs, the memory units retain pieces of information even when the sequences get really long.

### Bidirectional LSTM

For example, when you want to assign a sentiment score to a piece of text (say a customer review), the network can see the entire review text before assigning them a score. On the other hand, in a task such as predicting the next word given previous few typed words, the network does not have access to the words in the future time steps while predicting the next word.

These two types of tasks are called offline and online sequence processing respectively.

Now, there is a neat trick you can use with offline tasks — since the network has access to the entire sequence before making predictions, why not use this task to make the network ‘look at the future elements in the sequence’ while training, hoping that this will make the network learn better?
This is the idea exploited by what is called bidirectional RNNs.
By using bidirectional RNNs, it is almost certain that you’ll get better results. However, bidirectional RNNs take almost double the time to train since the number of parameters of the network increase. Therefore, you have a tradeoff between training time and performance. The decision to use a bidirectional RNN depends on the computing resources that you have and the performance you are aiming for.
